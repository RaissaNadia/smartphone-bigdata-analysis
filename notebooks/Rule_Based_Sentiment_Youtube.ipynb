# RULE BASED SENTIMENT

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Youtube_sentiment.csv")
df
```

```python
positif = {

    # umum
    "bagus",
    "mantap",
    "keren",
    "oke",
    "baik",
    "stabil",
    "cepat",
    "halus",
    "lancar",
    "responsif",

    # harga
    "murah",
    "worth",
    "worthit",
    "terjangkau",
    "hemat",

    # performa
    "kencang",
    "ngebut",
    "smooth",
    "powerful",
    "responsif",

    # gaming
    "gaming",
    "fps",
    "stabil",
    "anti_lag",

    # kamera
    "tajam",
    "jernih",
    "detail",
    "bagusnya",
    "bening",

    # baterai
    "awet",
    "irit",
    "tahan",
    "fastcharge",

    # layar
    "amoled",
    "cerah",
    "refresh",

    # fitur
    "lengkap",
    "premium",
    "unggul",

    # rekomendasi
    "rekomendasi",
    "recommended",
    "favorit"
}
```

```python
negatif = {

    # umum
    "jelek",
    "buruk",
    "parah",
    "kecewa",

    # performa
    "lemot",
    "lag",
    "ngelag",
    "patah",
    "freeze",
    "stuck",

    # suhu
    "panas",
    "overheat",

    # baterai
    "boros",
    "drop",
    "cepat_habis",

    # kamera
    "blur",
    "buram",
    "gelap",
    "noise",

    # layar
    "ghost_touch",
    "pecah",
    "retak",

    # sistem
    "bug",
    "error",
    "crash",

    # harga
    "mahal",
    "kemahalan",

    # kualitas
    "rusak",
    "cacat",
    "zonk",

    # pengalaman
    "nyesel",
    "mengecewakan"
}
```

```python
def sentiment_rule(text):

    skor = 0

    kata = str(text).split()

    for k in kata:

        if k in positif:
            skor += 1

        elif k in negatif:
            skor -= 1

    if skor > 0:
        return "Positif"

    elif skor < 0:
        return "Negatif"

    else:
        return "Netral"
```

```python
df["Sentiment_Rule"] = (
    df["Comment_Final"]
    .apply(sentiment_rule)
)

df["Sentiment_Rule"].value_counts()
```

```python
persen = (
    df["Sentiment_Rule"]
    .value_counts(normalize=True)
    * 100
).round(2)

print(persen)
```
